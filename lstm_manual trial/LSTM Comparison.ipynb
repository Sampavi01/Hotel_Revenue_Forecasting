This section compares **two LSTM hybrid architectures** used across all Revenue Centres and meal periods, focusing on **model structure, tuning behavior, and observed performance patterns**.

---

## 🧠 Model Definitions

### 🔹 Model A — **Deeper Stacked LSTM (Improved Architecture)**
```python
LSTM → LSTM → Dense(units/2) → Dense(units/4) → Output






**Key Characteristics**
- 2 stacked LSTM layers (captures deeper temporal dependencies)
- Funnel-shaped dense layers for feature compression
- Higher representational capacity
- Slightly increased risk of overfitting on sparse / zero-heavy data

---

### 🔹 Model B — **Shallow LSTM (Baseline Architecture)**

**Architecture**
LSTM → Dense(units/2) → Output

yaml
Copy code

**Key Characteristics**
- Single LSTM layer
- Fewer trainable parameters
- More stable on low-variance or near-constant time series
- Lower capacity to capture complex temporal patterns

---

## 📈 Hyperparameter Search Behavior

| Aspect | Model A (Stacked) | Model B (Single) |
|------|------------------|------------------|
| Typical Units | 64 – 128 | 128 |
| Learning Rate | 0.0005 – 0.01 | Mostly 0.01 |
| Ridge Alpha | 1 – 100 | 1 – 100 |
| Ensemble Weights | LSTM ≈ 0.45 – 0.53 | LSTM ≈ 0.40 – 0.53 |
| CV RMSE Stability | Slightly higher variance | More consistent |

---

## 🧪 Test Performance Patterns

### ✅ Where **Model A** Performs Better
- Revenue Centres **2 & 3**
- Meal periods with **non-zero, volatile demand**
- Scenarios where **temporal dependencies** are important
- Often achieves **lower RMSE** than Model B
- Produces **more balanced LSTM–Ridge ensemble weights**

### ⚠️ Where **Model A** Struggles
- Revenue Centre **4** (zero / constant series)
- Extremely sparse breakfast or lunch segments
- Inflated **MAPE / sMAPE** due to near-zero actual values

---

### ✅ Where **Model B** Performs Better
- Revenue Centre **4** (all meal periods)
- Stable or deterministic revenue patterns
- Produces **perfect RMSE = 0** when data is constant
- Less sensitivity to learning rate changes

### ⚠️ Where **Model B** Is Limited
- Higher RMSE on complex dinner/lunch patterns
- Misses long-term temporal dependencies
- Slightly worse generalization on volatile revenue centres

---

## 📉 Error Metric Observations

| Metric | Observation |
|------|------------|
| RMSE / MAE | Reliable for model comparison |
| MAPE | Explodes when actual values ≈ 0 |
| sMAPE | More stable but still inflated on sparse data |
| CV vs Test | CV RMSE generally aligns with test RMSE |

> ⚠️ **MAPE is not a valid metric for zero-heavy revenue series** and should not be used for model selection in this dataset.

---

## 🧩 Ensemble Behavior (LSTM + Ridge)

**Ridge Dominates (>50% weight) When**
- Revenue patterns are linear or near-constant
- LSTM tends to overfit

**LSTM Weight Increases When**
- Strong temporal patterns exist
- Revenue variance is high

---

## 🏁 Final Takeaways

### 🔹 Model A — Stacked LSTM
- ✔ Better for complex, dynamic revenue patterns
- ❌ Overkill for sparse or zero-demand centres

### 🔹 Model B — Single LSTM
- ✔ More robust and stable
- ❌ Limited temporal expressiveness

---

## ✅ Recommended Strategy

- Use **Model A** for Revenue Centres **2, 3, and 5**
- Use **Model B or Ridge-only** for Revenue Centre **4**
- Replace **MAPE** with **MAE / RMSE / WAPE**
- Apply **zero-demand filtering or epsilon smoothing** before computing percentage-based errors

---

📌 *This comparison is based on empirical results from random search cross-validation and ensemble evaluation across all Revenue Centres and meal periods.*
