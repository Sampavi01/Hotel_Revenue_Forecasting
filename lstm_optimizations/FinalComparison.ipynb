This section presents a complete evaluation of multiple **LSTM–Dense hybrid architectures** across different **Revenue Centres** and **meal periods**.  
Models include shallow, moderate, and deep LSTM stacks combined with varying numbers of Dense (hidden) layers.

All models were evaluated using **time-series cross-validation RMSE**, where **lower RMSE indicates better forecasting performance**.

---

## 🏢 Revenue Centre 1

### 🍳 Breakfast — Model Performance Summary

The best-performing architecture for Breakfast revenue at Revenue Centre 1 is **4×LSTM + 3 Dense**, achieving the lowest CV RMSE of **950.7**.

Key observations:
- Deep LSTM stacks significantly improve performance.
- Additional Dense layers help capture complex nonlinear relationships.
- Shallow architectures underperform due to limited temporal modeling.

**Conclusion:**  
Breakfast revenue at Revenue Centre 1 exhibits strong long-term temporal dependencies and benefits from deep LSTM architectures with expanded Dense layers.

---

### 🍽 Lunch — Model Performance Summary

For Lunch revenue at Revenue Centre 1, the optimal architecture is **2×LSTM + 2 Dense**, with a CV RMSE of **789.76**.

Key observations:
- Moderate LSTM depth achieves the best bias–variance tradeoff.
- Deeper LSTM stacks do not yield additional performance gains.
- Dense expansion beyond two layers provides no benefit.

**Conclusion:**  
Lunch revenue patterns are stable and structured, requiring moderate temporal modeling rather than deep memory networks.

---

### 🍽 Dinner — Model Performance Summary

The lowest CV RMSE for Dinner revenue at Revenue Centre 1 is achieved by **4×LSTM + 3 Dense**, with a value of **1561.12**.

Key observations:
- Dinner revenue is highly volatile.
- Deep LSTM architectures capture long-range dependencies more effectively.
- Shallow models fail to generalize under high variability.

**Conclusion:**  
Dinner revenue at Revenue Centre 1 requires deep temporal modeling to handle its complex and noisy behavior.

---

## 🏢 Revenue Centre 2

### 🍳 Breakfast — Model Performance Summary

For Breakfast revenue at Revenue Centre 2, the best-performing model is **2×LSTM + 2 Dense**, achieving a CV RMSE of **331.4**.

Key observations:
- Revenue patterns are highly predictable.
- Deeper LSTM stacks introduce unnecessary complexity.
- Moderate-depth architectures generalize best.

**Conclusion:**  
Breakfast revenue at Revenue Centre 2 exhibits short-to-medium temporal dependencies and benefits from simpler LSTM architectures.

---

### 🍽 Lunch — Model Performance Summary

The optimal architecture for Lunch revenue at Revenue Centre 2 is **2×LSTM + 2 Dense**, with a CV RMSE of **415.92**.

Key observations:
- Balanced temporal and linear patterns are present.
- Additional LSTM depth does not improve forecasting accuracy.
- Over-parameterized models increase variance.

**Conclusion:**  
Lunch revenue at Revenue Centre 2 is stable, and moderate-depth LSTM architectures provide the most reliable forecasts.

---

### 🍽 Dinner — Model Performance Summary

For Dinner revenue at Revenue Centre 2, the lowest RMSE of **887.5** is obtained using **4×LSTM + 3 Dense**.

Key observations:
- Dinner demand exhibits complex temporal behavior.
- Deep LSTM stacks marginally but consistently improve performance.
- Dense expansion helps refine nonlinear trends.

**Conclusion:**  
Dinner revenue at Revenue Centre 2 benefits from deeper temporal architectures due to increased demand variability.

---

## 📌 Cross-Centre Architectural Insights

Across both Revenue Centres, the following patterns emerge:
- **Moderate-depth architectures (2×LSTM + 2 Dense)** perform best for **stable and predictable revenue streams**.
- **Deep architectures (4×LSTM + 3 Dense)** outperform others for **highly volatile meal periods**, particularly Dinner.
- No single architecture is universally optimal across all scenarios.

## ✅ Final Modeling Strategy

Based on experimental results, the following strategy is recommended:

- **Breakfast & Lunch:**  
  Use **2×LSTM + 2 Dense** architectures for stability and generalization.

- **Dinner:**  
  Use **4×LSTM + 3 Dense** architectures to handle higher volatility.

If a single global model is required, **4×LSTM + 3 Dense** provides the most robust performance across diverse conditions.
